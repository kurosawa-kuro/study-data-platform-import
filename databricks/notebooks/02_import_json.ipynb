# 02 Import JSON

`events.json` を読み込み、Delta Table として保存する。

```python
events_path = "/FileStore/data/events.json"
```

```python
df_events = spark.read.option("multiline", True).json(events_path)
display(df_events)
```

```python
df_events.createOrReplaceTempView("events_view")
spark.sql("CREATE DATABASE IF NOT EXISTS study_import")
spark.sql("USE study_import")
df_events.write.mode("overwrite").saveAsTable("events")
```
