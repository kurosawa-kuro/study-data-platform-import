# 03 Import Parquet

`products.parquet` を読み込み、Delta Table として保存する。

```python
products_path = "/FileStore/data/products.parquet"
```

```python
df_products = spark.read.parquet(products_path)
display(df_products)
```

```python
df_products.createOrReplaceTempView("products_view")
spark.sql("CREATE DATABASE IF NOT EXISTS study_import")
spark.sql("USE study_import")
df_products.write.mode("overwrite").saveAsTable("products")
```
